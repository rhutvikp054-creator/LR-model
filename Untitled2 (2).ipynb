```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
  df=pd.read_csv("/content/Advertising.csv",index_col=0)
```

```python
df.head()
```

```python
df
```

```python
x=df.drop('Sales',axis=1)
y=df['Sales']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
from sklearn.linear_model import LinearRegression
```

```python

LR=LinearRegression()
```

```python

lr=LR.fit(X_train,y_train)
```

```python
lr.intercept_
```

```python
lr.coef_
```

```python
lr.score(X_train,y_train)
```

```python
lr.score(X_test,y_test)
```

```python
lr.predict([[5,4,2]])
```

```python
from sklearn.model_selection import cross_val_score
```

```python
cross_val_score(lr,x,y,cv=5)
```

```python
cross_val_score(lr,x,y,cv=5).mean()
```

```python
import joblib

# Save the model to a file
joblib.dump(lr, 'linear_regression_model.sav')
print("Model saved as linear_regression_model.sav")
```

```python
import joblib

# Load the model from the file
loaded_lr = joblib.load('linear_regression_model.sav')
print("Model loaded successfully!")
print(f"Loaded model intercept: {loaded_lr.intercept_}")
print(f"Loaded model coefficients: {loaded_lr.coef_}")
```

```python
loaded_lr.predict([[5,4,2]])
```

```python
import streamlit as st
import joblib
import pandas as pd

# Load the trained model
loaded_lr = joblib.load('linear_regression_model.sav')

st.title('Sales Prediction with Linear Regression')
st.write('Enter the advertising budgets for TV, Radio, and Newspaper to predict sales.')

# Input fields for features
tv = st.slider('TV Advertising Budget ($)', 0.0, 300.0, 150.0)
radio = st.slider('Radio Advertising Budget ($)', 0.0, 50.0, 25.0)
newspaper = st.slider('Newspaper Advertising Budget ($)', 0.0, 120.0, 30.0)

# Create a DataFrame for prediction
# The order of features must match the order used during training (TV, Radio, Newspaper)
input_data = pd.DataFrame([[tv, radio, newspaper]],
                            columns=['TV', 'Radio', 'Newspaper'])


# Make prediction
if st.button('Predict Sales'):
    prediction = loaded_lr.predict(input_data)
    st.success(f'Predicted Sales: {prediction[0]:.2f} units')

st.write('To run this application:')
st.code('streamlit run app.py')
```
